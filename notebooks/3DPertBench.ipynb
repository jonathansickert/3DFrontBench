```python
import json
import random

min_k = 0
max_k = 3
min_rotation = 30
min_axis=1
max_axis=2

scale = [0.67, 0.8, 1.25, 1.5]
translation = [0.67, 0.8, 1.25, 1.5]
```

```python
with open("/home/jonathansickert/git/3DFrontBench/assets/objects.json") as f:
    objects = json.load(f)
```

```python
def sample_perturbation(obj: dict) -> list[str]:
    perturbation = random.choice(obj["Perturbation"])
    if perturbation == "rotation":
        return [perturbation, obj["Rotation"], random.randint(min_rotation, obj["Rotation Limit"]) * random.choice([-1, 1])]
    
    if perturbation == "scale":
        return [perturbation, random.choice(scale)]
    
    if perturbation == "placement":
        selected = random.sample(obj["Placement"], k=random.randint(min_axis, max_axis))
        placement = []
        for s in selected:
            placement.append(s)
            placement.append(random.choice(translation) * random.choice([-1, 1]))
        return [perturbation, *placement]
    
    if perturbation == "count":
        return [perturbation]

for i in range(10):
    perturbed = {}
    for scene, objs in objects.items():
        selected = random.sample(objs, k=random.randint(min_k, max_k))
        perturbations = {obj["Object"] : sample_perturbation(obj) for obj in selected}
        perturbed[scene] = perturbations


    with open(f"/home/jonathansickert/git/3DFrontBench/perturbations/perturbations_{i}.json", "w") as f:
        json.dump(perturbed, f, indent=2)
```

```python
for variant in range(0, 5):
    perturbed = {}
    for scene, objs in objects.items():
        selected = random.sample(objs, k=random.randint(min_k, max_k))
        selected = [obj["Object"] for obj in selected]

        not_selected = [obj["Object"] for obj in objs if obj["Object"] not in selected]


        perturbed[scene] = {obj : {"perturbation_type" : "count", "args" : None} for obj in selected}
        perturbed[scene].update({obj : {"perturbation_type" : "none", "args" : None} for obj in not_selected})

    with open(f"/home/jonathansickert/git/3DFrontBench/perturbations_count/perturbations_{variant}.json", "w") as f:
        json.dump(perturbed, f, indent=2)
```
